```python
from nltk.tokenize import word_tokenize

text = "I love machine learning and deep learning"
words = word_tokenize(text)

words
```

```python
words.count("learning")
```

```python
from nltk.probability import FreqDist

fdist = FreqDist(words)
fdist.most_common(5)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered = [w for w in words if w.lower() not in stop_words]
filtered
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered = [w for w in words if w.lower() not in stop_words]
filtered
```

```python
from nltk import bigrams

list(bigrams(filtered))
```

```python
from nltk.collocations import BigramCollocationFinder
from nltk.metrics import BigramAssocMeasures

finder = BigramCollocationFinder.from_words(filtered)

bigram_measures = BigramAssocMeasures()

finder.nbest(bigram_measures.pmi, 5)
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

text = "I love this product. It is amazing!"

score = sia.polarity_scores(text)

print("Kết quả phân tích cảm xúc:")
print(score)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

text = "Data science is amazing. I love data science and AI."

# Tokenize
words = word_tokenize(text.lower())

# Remove stopwords
stop_words = set(stopwords.words('english'))
filtered = [w for w in words if w not in stop_words]

# Frequency
fdist = FreqDist(filtered)

print("Top 3 từ xuất hiện nhiều nhất:")
print(fdist.most_common(3))
```
